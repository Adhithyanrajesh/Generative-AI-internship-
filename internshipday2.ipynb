```python
import pandas as  pd

from sklearn.preprocessing import LabelEncoder

df=pd.read_csv('/content/preprocessing_practice_dataset.csv')

encoded_data=df.copy()

label_encoder=LabelEncoder()

encoded_data['Gender_Label']=label_encoder.fit_transform(encoded_data['Gender'])

one_hot_encoded=pd.get_dummies(encoded_data['Department'],prefix='Department')
encoded_data=pd.concat([encoded_data,one_hot_encoded],axis=1)
perfomance_mapping={1:'First',2:'Second',3:'Third',4:'four',5:'five'}

encoded_data['Performance_label']=encoded_data['Performance_Rating'].map(perfomance_mapping)

encoded_data


```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
titanic_data=pd.read_csv('/content/preprocessing_practice_dataset.csv')

numerical_coloumns = ['Age','Experience_Years','Monthly_Salary','Performance_Rating']

titanic_data.dropna(subset=numerical_coloumns,inplace=True)
min_max_scaler=MinMaxScaler()
standard_scaler=StandardScaler()

titanic_data['Age_Standard']=standard_scaler.fit_transform(titanic_data[['Age']])

titanic_data['Experience_Years_standard']=standard_scaler.fit_transform(titanic_data[['Experience_Years']])

titanic_data['Monthly_Salary_Standard']=standard_scaler.fit_transform(titanic_data[['Monthly_Salary']])

titanic_data['Performance_Rating_Standard']=standard_scaler.fit_transform(titanic_data[['Performance_Rating']])

titanic_data['Age_MinMax']=min_max_scaler.fit_transform(titanic_data[['Age']])

titanic_data['Experience_Years_MinMax']=min_max_scaler.fit_transform(titanic_data[['Experience_Years']])

titanic_data['Monthly_Salary_MinMax']=min_max_scaler.fit_transform(titanic_data[['Monthly_Salary']])

titanic_data['Performance_Rating_MinMax']=min_max_scaler.fit_transform(titanic_data[['Performance_Rating']])
print(titanic_data[['Age_Standard','Experience_Years_standard','Monthly_Salary_Standard','Performance_Rating_Standard']])

print(titanic_data[['Age_MinMax','Experience_Years_MinMax','Monthly_Salary_MinMax','Performance_Rating_Standard']])
```
